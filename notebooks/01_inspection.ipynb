Inpection du dataset justifiant les décision de nettoyage au niveau de scripts/nettoyage.py

```python
import pandas as pd
import numpy as np

# Charger le fichier Excel
df = pd.read_excel("../data/Base Kidney diseases.xlsx", sheet_name="kidney_disease")

# Dimensions
print("Dimensions:", df.shape)
print("Nombre de lignes:", df.shape[0])
print("Nombre de colonnes:", df.shape[1])
```

```python
# Aperçu
pd.set_option('display.max_columns', None)
print(df.head(10))
```

```python
# Informations sur les types et les manquants
df.info()
```

```python
# Statistiques descriptives (toutes colonnes)
print(df.describe(include='all'))
```

```python
# Nombre de manquants par colonne
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'manquants': missing,
    'pourcentage': missing_pct
}).sort_values('pourcentage', ascending=False)

print(missing_df)
```

```python
# Pour chaque colonne, afficher les valeurs uniques
for col in df.columns:
    print(f"--- {col} ---")
    print(f"Type: {df[col].dtype}")
    print(f"Valeurs uniques: {df[col].nunique()}")
```

```python
# Cellules converties en date par Excel, par colonne
import datetime as dt
est_date = lambda v: isinstance(v, (dt.datetime, pd.Timestamp))
dates = df.apply(lambda s: int(s.map(est_date).sum()))
print(dates[dates > 0])
print("Total :", dates.sum())
```

```python
# Marqueur '?' utilisé comme valeur manquante
interrogations = (df == "?").sum()
print(interrogations[interrogations > 0])
```

```python
# Distribution de la cible
print(df['classification'].value_counts(dropna=False))
print()
print(df['classification'].value_counts(normalize=True, dropna=False) * 100)
```

```python
# Doublons exacts
print("Doublons exacts:", df.duplicated().sum())

# Doublons sur l'id
print("Doublons id:", df['id'].duplicated().sum())
```
